```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df=pd.read_csv("expense_receipts_fraud_detection_2000.csv")
```

```python
display(df.head())
```

```python
display(df.info())
```

```python
print(df.isnull().sum())
```

```python
display(df.describe())
```

```python
# 1. Target Variable: Proportion of Fraud
plt.figure(figsize=(6, 6))
fraud_counts = df['is_fraud'].value_counts()
plt.pie(fraud_counts, labels=['Normal (0)', 'Fraud (1)'], autopct='%1.1f%%', 
        startangle=90, colors=['#66b3ff', '#ff9999'])
plt.title('Proportion of Fraudulent Transactions')
plt.show()

# 2. Expense Categories Distribution
plt.figure(figsize=(10, 6))
# Sort the categories by count
order = df['category'].value_counts().index
sns.countplot(data=df, y='category', order=order, palette='viridis')
plt.title('Transaction Count by Category')
plt.xlabel('Number of Transactions')
plt.ylabel('Expense Category')
plt.show()

# 3. Top 10 Locations
plt.figure(figsize=(10, 6))
top_locations = df['location'].value_counts().head(10).index
sns.countplot(data=df, y='location', order=top_locations, palette='magma')
plt.title('Top 10 Transaction Locations')
plt.xlabel('Number of Transactions')
plt.ylabel('City / Location')
plt.show()
```

```python
# 1. Fraud by Expense Category
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='category', hue='is_fraud', palette='Set1', order=order)
plt.title('Fraud Status across Expense Categories')
plt.xticks(rotation=45)
plt.xlabel('Category')
plt.ylabel('Count')
plt.legend(title='Is Fraud', labels=['No', 'Yes'])
plt.tight_layout()
plt.show()

```
